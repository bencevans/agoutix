# Summarise Contributions to a Project

```python
from agoutix.public import AgoutiPublicAPI, AuthJWT
from agoutix import Agouti

agouti = Agouti("EMAIL", "PASSWORD")
public_api = AgoutiPublicAPI(AuthJWT(agouti.token))


project_ids = [
    "aaaaaaaa-bbbb-cccc-dddd-eeeeeeeeeeee",
]
```

```python
def enumerate_observations(project_id: str):
  page = 1
  all_observations = []

  while True:
    incoming_observations = []
    while True:
      try:
        incoming_observations = agouti.list_project_observations(project_id, page_offset=page)
        break
      except Exception as e:
        print(f"Error on page {page} of project {project_id}: {e}")
        print("Retrying...")
    # incoming_observations = agouti.list_project_observations(project_id, page_offset=page)
    all_observations.extend(incoming_observations)

    if len(incoming_observations) == 0:
      # print(f"breaking at page {page}")
      break
    else:
      page += 1

  return all_observations



observations = enumerate_observations(project_ids[0])
# len()
```

```python
from rich import print
from tqdm.auto import tqdm

all_observations = []
for project_id in tqdm(project_ids):
  all_observations.extend(enumerate_observations(project_id))
```

```python
names = [f"{ob.attributes.created_by_first_name} {ob.attributes.created_by_last_name}" for ob in all_observations]

len(names)
```

```python
sequence_ids = {obs.attributes.sequence_id for obs in all_observations}
len(sequence_ids)
```

```python
observation_ids = {obs.id for obs in all_observations}
len(observation_ids)
```

```python
observations_by_id = {obs.id: obs for obs in all_observations}
names_by_observation_id = {obs.id: f"{obs.attributes.created_by_first_name} {obs.attributes.created_by_last_name}" for obs in all_observations}
len(observations_by_id)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

counts = pd.Series(names_by_observation_id).value_counts().sort_values(ascending=True)
ax = counts.plot(kind="barh", grid=True)
for patch in ax.patches:
    ax.text(patch.get_width() + counts.max() * 0.01, patch.get_y() + patch.get_height() / 2, str(int(patch.get_width())), va="center", ha="left")
plt.title("Number of observations contributed by each user")
plt.tight_layout()
```
